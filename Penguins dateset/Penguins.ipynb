```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 

penguins = sns.load_dataset("penguins")
penguins
```

```python
#  species count 
sns.countplot(data=penguins,x='species')
plt.title('species count ')
plt.show()
```

```python
# 2. Island Count
sns.countplot(data=penguins,x='island')
plt.title('island count ')
plt.show()
```

```python
# . Sex Count
sns.countplot(data=penguins,x='sex')
plt.title('sex count ')
plt.show()
```

```python
# Bill Length Histogram
plt.figure(figsize=(10,8))
sns.histplot(data=penguins,x='bill_length_mm',bins=20,color='red',kde=True)
plt.title('bill_length distribution ')
plt.show()
```

```python
# Bill Depth KD
plt.figure(figsize=(10,8))
sns.kdeplot(data=penguins,x='bill_depth_mm',fill=True)
plt.title('bill depth  distribution ')
plt.show()
```

```python
# Boxplot of Body Mass
plt.Figure(figsize=(10,7))
sns.boxplot(data=penguins,y="body_mass_g")
plt.title("body mass distribution")
plt.show()
```

```python
# Violinplot by Species
plt.Figure(figsize=(10,7))
sns.violinplot(data=penguins,x="species",y="body_mass_g")
plt.title("boy mass by species ")
plt.show()
```

```python
# Scatter Bill Length vs Bill Depth
plt.figure(figsize=(10,8))
sns.scatterplot(data=penguins,x='bill_length_mm',y='bill_depth_mm')
plt.title('relationship between bill length vs bill depth ')
plt.show()
```

```python
# plt.show()
# 9. Scatter + Hue = Species
plt.figure(figsize=(10,8))
sns.scatterplot(data=penguins,x='bill_length_mm',y='bill_depth_mm',hue='species')
plt.title('relationship between bill length vs bill depth by species ')
plt.show()
```

```python
#Scatter + Style = Sex 
plt.figure(figsize=(10,8))
sns.scatterplot(data=penguins,x='bill_length_mm',y='bill_depth_mm',hue='species',style='sex')
plt.title('relationship between bill length vs bill depth by species ')
plt.show()
```

```python
# . Pairplot
sns.pairplot(data=penguins,hue='species')
plt.title('distribution of a numerical value ')
plt.show()
```

```python
# 2. PairGrid
g = sns.PairGrid(data=penguins)
g.map_upper(sns.scatterplot)
g.map_diag(sns.histplot)
g.map_lower(sns.kdeplot)
```

```python
g = sns.PairGrid(
    data=penguins,
    vars=["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
)

g.map(sns.scatterplot)
plt.show()
```

```python
# Heatmap

corr = penguins.corr(numeric_only=True)

sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
corr = penguins.select_dtypes("number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# 14. Clustermap

corr = penguins.corr(numeric_only=True)

sns.clustermap(corr,annot=True,cmap='coolwarm')
```

```python
# Jointplot
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    kind="scatter"
)

plt.show()
```

```python
# KDE version: of aq joint plot 
sns.jointplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    kind="kde"
)

plt.show()
```

```python
# 16. FacetGrid by Island 
g = sns.FacetGrid(
    penguins,
    col="island"
)

g.map_dataframe(
    sns.scatterplot,
    x="bill_length_mm",
    y="bill_depth_mm"
)

plt.show()
```

```python
# Relplot

"""
Bhai, relplot() ka main purpose hai numerical variables ke beech relationship ko visualize karna, especially jab tumhe multiple categories ya multiple subplots ke saath relationship compare karna ho.
    """
sns.relplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    col="island"
)

plt.show()
```

```python
# . Catplot
sns.catplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    hue="sex",
    kind="box"
)

plt.show()
```

```python
# 19. Missing Values Visualization
#  barplot --Barplot ✅ → category vs single numerical value comparison
# Simple missing-value count:
missing = penguins.isnull().sum()

sns.barplot(
    x=missing.index,  #Column names:
    y=missing.values# Har column ke missing values ki count
)

plt.xticks(rotation=45) # agar xticks nhi use karunga to overlap ho jata hai labels 
plt.ylabel("Missing Values")  
plt.title("Missing Values by Column")
plt.show()
```

```python
# Heatmap se visualize:

#  heatmap batata hai ki Batata hai:

# Missing values dataset me exactly kahan hain?
"""
Barplot → Missing values ka count 📊
Heatmap → Missing values ka location/pattern 🔥
    """


sns.heatmap(
    penguins.isnull(),
    cbar=False  #Heatmap ke side me jo color scale/legend aata hai, usko hide kar do.
)

plt.title("Missing Values Heatmap")
plt.show()
```

```python
# 0. Dashboard 🔥
fig, ax = plt.subplots(2, 2, figsize=(14, 10))

# 1. Species count
sns.countplot(
    data=penguins,
    x="species",
    ax=ax[0, 0]
)
ax[0, 0].set_title("Species Count")

# 2. Body Mass by Species
sns.boxplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    ax=ax[0, 1]
)
ax[0, 1].set_title("Body Mass by Species")

# 3. Bill Length vs Bill Depth
sns.scatterplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    ax=ax[1, 0]
)
ax[1, 0].set_title("Bill Length vs Bill Depth")

# 4. Body Mass Distribution
sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=20,
    ax=ax[1, 1]
)
ax[1, 1].set_title("Body Mass Distribution")

plt.tight_layout()
plt.show()
```
