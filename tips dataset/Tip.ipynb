```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt 

tips = sns.load_dataset('tips') 
tips.head()
```

```python
# Question 2

# Dataset ki information nikalo

tips.info()
tips.describe()
tips.shape
tips.columns
tips.isnull
```

```python
# Total Bill ka histogram banao.---- univariate 
"""
Purpose--

Distribution dekhna.
    """

sns.histplot(data=tips,x='total_bill')
plt.show()
```

```python
# Bins =20
sns.histplot(data=tips,x='total_bill',bins=20)
plt.show()
```

```python
# KDE bhi add karo.
# sns.kdeplot(data=tips,x="total_bill")
sns.histplot(data=tips, x="total_bill", kde=True)
```

```python
# Rug Plot add karo.

"""
Purpose

Har observation ko chhoti line se dikhata hai.
    """

sns.kdeplot(data=tips,x="total_bill")
sns.rugplot(data=tips,x="total_bill")
```

```python
# ECDF Plot----Purpose
#----------> Cumulative distribution.

sns.ecdfplot(data=tips,x="total_bill")
```

```python
# Box Plot--

"""Purpose

---Median, Quartiles aur Outliers dikhata hai.
    """
sns.boxplot(data=tips,x="total_bill")
```
