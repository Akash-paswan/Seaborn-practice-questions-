```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt 

tips = sns.load_dataset('tips') 
tips.head()
```

```python
# Question 2

# Dataset ki information nikalo

tips.info()
tips.describe()
tips.shape
tips.columns
tips.isnull
```

```python
# Total Bill ka histogram banao.---- univariate 
"""
Purpose--

Distribution dekhna.
    """

sns.histplot(data=tips,x='total_bill')
plt.show()
```

```python
# Bins =20
sns.histplot(data=tips,x='total_bill',bins=20)
plt.show()
```

```python
# KDE bhi add karo.
# sns.kdeplot(data=tips,x="total_bill")
sns.histplot(data=tips, x="total_bill", kde=True)
```

```python
# Rug Plot add karo.

"""
Purpose

Har observation ko chhoti line se dikhata hai.
    """

sns.kdeplot(data=tips,x="total_bill")
sns.rugplot(data=tips,x="total_bill")
```

```python
# ECDF Plot----Purpose
#----------> Cumulative distribution.

sns.ecdfplot(data=tips,x="total_bill")
```

```python
# Box Plot--

"""Purpose

---Median, Quartiles aur Outliers dikhata hai.
    """
sns.boxplot(data=tips,y="total_bill")
```

```python
# boxen plot ----

#  purpose --Large datasets me distribution ko aur detail me dikhata ha

sns.boxenplot(data=tips,y='total_bill')
```

```python
# violinplot ---- Boxplot + KDE

sns.violinplot(data=tips,y="total_bill")
```

```python
# Strip plot -->Har data point dikhata hai.


sns.stripplot(data=tips,y="total_bill")
```

```python
#  swarm plot -->Purpose

# Points overlap nahi hote.

sns.swarmplot(data=tips,y="total_bill")
```

```python
#  count plot 

sns.countplot(data=tips,x="day",hue="sex")
```

```python
#  level 3 bivariate analysis 
# Q15 Scatter Plot

# ---->2 Numerical variables ka relationship.


# sns.scatterplot(data=tips ,x="total_bill",y="tip")


# hue  -Color se categories dikhata hai.

sns.scatterplot(data=tips ,x="total_bill",y="tip",hue="sex")

```

```python
#  style --> Marker shape change hoti hai

sns.scatterplot(data=tips ,x="total_bill",y="tip",hue="sex",style="smoker")
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    size="size"
)
```

```python
#  line plot ---> Trend dekhna.
sns.lineplot(
    data=tips,
    x="size",
    y="tip"
)
```

```python
#  bar plot --> Default me Average dikhata hai.
sns.barplot(
    data=tips,
    x="day",
    y="tip"
)
```

```python
#  point plot -->  Category-wise average ko points aur lines se dikhata hai.
sns.pointplot(
    data=tips,
    x="day",
    y="tip"
)
```

```python
#  regplot ---Regression line dikhata hai.
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

```python
#  resid plot ---Regression errors (residuals) dikhata hai.
sns.residplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

```python
#  multivariate 

#  pair plot ----.Sabhi numerical columns ka relationship.
sns.pairplot(tips)
```

```python
#  pair grid ----Pairplot ka customizable version. 
g=sns.PairGrid(tips)
g.map_upper(sns.scatterplot)
g.map_diag(sns.histplot)
g.map_lower(sns.boxplot)
```

```python
#  heatmap ---Correlation matrix

sns.heatmap(tips.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")
```

```python
pivot = tips.pivot_table(
    values="total_bill",
    index="day",
    columns="sex"
)

sns.heatmap(pivot, annot=True)
```

```python
#  clustermap ----Correlation + Hierarchical clustering.
sns.clustermap(
    tips.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)
```

```python
#  figure level function 
"""
Purpose

Scatter/Line figure-level plot.
    """
    
# sns.relplot(data=tips,kind='scatter',x='total_bill',y="tip",hue='sex')    
sns.relplot(data=tips,kind='line',x='total_bill',y="tip",hue='sex')    
```

```python
#  catplot --->Purpose

# Figure-level categorical plots.

sns.catplot(
    data=tips,
    x="day",
    y="tip",
    kind="box"
)
```

```python
#  displot -->Distribution plot.\

sns.displot(data=tips,kind="hist",x='total_bill')
```

```python
#  joint plot --->Scatter + Histograms ek saath.
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="scatter"
)
```

```python
#  facetgrid --> Har category ke liye alag graph. , hm koi bhi graph bna skte hai 

g = sns.FacetGrid(tips, col="sex")
g.map_dataframe(sns.histplot,
                x="total_bill",
                y="tip")
```

```python
# FacetGrid + Histogram

g = sns.FacetGrid(tips, col="sex")
g.map_dataframe(sns.histplot,
                x="total_bill")
```

```python

#  facetgrid + kde 
g = sns.FacetGrid(tips, col="sex")
g.map_dataframe(sns.kdeplot,
                x="total_bill")
```
