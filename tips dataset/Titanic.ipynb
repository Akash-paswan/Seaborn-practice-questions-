```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 

titanic = sns.load_dataset('titanic')
titanic
```

```python
# Male vs Female Countplot
plt.Figure(figsize=(10,6))
sns.countplot(data=titanic,x='sex')
plt.title("male vs female passenger ")
plt.show()
```

```python
# 4. Survival Countplot
plt.figure(figsize=(6,4))

sns.countplot(data=titanic, x="survived")

plt.title("Survival Count")
plt.show()
```

```python
# Survival by Gender
plt.figure(figsize=(6,4))

sns.countplot(
    data=titanic,
    x="sex",
    hue="survived"
)

plt.title("Survival by Gender")
plt.show()
```

```python
# . Survival by Passenger Class 
plt.figure(figsize=(10,8))
sns.countplot(data=titanic,
              x='class',
             hue='survived' )
plt.title('survived by ps class ')
plt.show
```

```python
# 7. Age Distribution Histogram
plt.figure(figsize=(10,6))
sns.histplot(data=titanic,x="age",bins=25,kde=True)
```

```python
# Age KDE Plot
plt.figure(figsize=(10,6))
sns.kdeplot(data=titanic,x="age",fill = True)
plt.show
```

```python
# Fare Histogram
plt.Figure(figsize=(7,4))
sns.histplot(data=titanic,x='fare',bins=30)
```

```python
# 0. Fare Boxplot
plt.figure(figsize=(5,6))

sns.boxplot(
    data=titanic,
    y="fare"
)

plt.show()
```

```python
#  fare by passenger calss 
plt.figure(figsize=(7,5))

sns.boxplot(
    data=titanic,
    x="class",
    y="fare"
)

plt.show()
```

```python
#  age by passenger class
sns.boxplot(data=titanic,x="pclass",y="age")
```

```python
# Age vs Fare
sns.scatterplot(data=titanic,x='fare',y='age')
```

```python
# 14. Age vs Fare + Hue = Survived
sns.scatterplot(data=titanic,x='fare',y='age',hue='survived')
```

```python
# Heatmap Correlation
plt.figure(figsize=(10,7))
Corr =titanic.corr(numeric_only=True)

sns.heatmap(Corr,annot=True,cmap='coolwarm')
```

```python
# 17. Violinplot of Age by Sex
plt.Figure(figsize=(10,7))
sns.violinplot(data=titanic,x='sex',y='age')
```

```python
# . Swarmplot of Fare by Class
plt.figure(figsize=(7,5))

sns.swarmplot(
    data=titanic,
    x="class",
    y="fare"
)

plt.show()
```

```python
sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    kind="box",
    height=5,
    aspect=1.5
)

plt.show()
```

```python
#  complete Eda dashboard 
fig, ax = plt.subplots(2, 2, figsize=(14,10))

# 1
sns.countplot(
    data=titanic,
    x="survived",
    ax=ax[0,0]
)
ax[0,0].set_title("Survival Count")

# 2
sns.countplot(
    data=titanic,
    x="sex",
    hue="survived",
    ax=ax[0,1]
)
ax[0,1].set_title("Survival by Gender")

# 3
sns.histplot(
    data=titanic,
    x="age",
    kde=True,
    ax=ax[1,0]
)
ax[1,0].set_title("Age Distribution")

# 4
sns.boxplot(
    data=titanic,
    x="class",
    y="fare",
    ax=ax[1,1]
)
ax[1,1].set_title("Fare by Class")

plt.tight_layout()
plt.show()
```
