```python
import seaborn as sns
import matplotlib.pyplot as plt

diamonds = sns.load_dataset("diamonds")

diamonds.head()
```

```python
diamonds["color"].value_counts()
```

```python
diamonds.shape
diamonds.info()
diamonds.describe()
diamonds.isnull().sum()
diamonds.columns
```

```python
sns.kdeplot(
    data=diamonds,
    x="price",
    fill=True
)

plt.title("Price KDE")
plt.show()
```
