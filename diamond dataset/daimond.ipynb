```python
import seaborn as sns
import matplotlib.pyplot as plt

diamonds = sns.load_dataset("diamonds")

diamonds.head()
```

```python
diamonds["color"].value_counts()
```

```python
diamonds.shape
diamonds.info()
diamonds.describe()
diamonds.isnull().sum()
diamonds.columns
```

```python
sns.kdeplot(
    data=diamonds,
    x="price",
    fill=True
)

plt.title("Price KDE")
plt.show()
```

```python
sns.scatterplot(
    data=diamonds,
    x="carat",
    y="price"
)

plt.title("Carat vs Price")
plt.xlabel("Carat")
plt.ylabel("Price")
plt.show()
```

```python
corr = diamonds.select_dtypes("number").corr()

sns.heatmap(
    corr,
    annot=True
)

plt.title("Diamond Correlation Heatmap")
plt.show()
```

```python
cut_count = diamonds["cut"].value_counts(normalize=True) * 100

sns.barplot(
    x=cut_count.index,
    y=cut_count.values
)

plt.ylabel("Percentage")
plt.title("Diamond Percentage by Cut")
plt.show()
```

```python
sns.displot(
    data=diamonds,
    x="price",
    col="cut",
    kde=True
)
plt.show()
```
