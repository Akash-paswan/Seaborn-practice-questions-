```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 

sns.set_theme()
iris = sns.load_dataset('iris')
iris.head()
```

```python
# Species countplot

sns.countplot(data=iris,x='species')
plt.title("species count ")
```

```python
# Sepal Length Histogram

plt.Figure(figsize=(10,6))
sns.histplot(data=iris,x='sepal_length',color="green")
plt.title(" sepal length distribution ")
plt.show
```

```python
# Petal Width KDE
plt.figure(figsize=(10,7))
sns.kdeplot(data=iris,x='petal_width',fill=True,color='red')
plt.title("petal width distribution")
plt.show()
```

```python
# Boxplot of Sepal Length
plt.figure(figsize=(8,7))
sns.boxplot(data=iris,y='sepal_length')
```

```python
plt.figure(figsize=(10,8))
sns.violinplot(data=iris,x='species')
plt.title("species of a violinplot")
plt.show
```

```python
# Scatterplot Sepal Length vs Width
plt.figure(figsize=(10,8))
sns.scatterplot(data=iris,x='sepal_width',y='sepal_width')
plt.title('scatterplot of sepal length vs sepal width')
plt.show
```

```python
# Scatter + Hue = Species
plt.figure(figsize=(10,8))
sns.scatterplot(data=iris,x='sepal_width',y='sepal_width',hue="species")
plt.title('scatterplot of sepal length vs sepal width')
plt.show
```

```python
# Pairplot
sns.pairplot(iris,hue="species")
plt.show()
```

```python
# PairGrid

# Pairplot ka customizable version samjho.

g = sns.PairGrid(iris,hue='species')
g.map_upper(sns.histplot)
g.map_diag(sns.kdeplot)
g.map_lower(sns.scatterplot)
g.add_legend()
```

```python
# Correlation ko visualize karne ke liye:

corr = iris.corr(numeric_only=True)

sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title('coorelation heatmap ')
plt.show
```

```python
corr = iris.iloc[:, 0:4].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="plasma"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# /Clustermap
# 
# Heatmap + hierarchical clustering:
corr = iris.iloc[:, 0:4].corr()
sns.clustermap(corr,
    annot=True,
    cmap="coolwarm"
)
```

```python
# 12. Jointplot

# Do numerical variables ka relationship + distributions:

sns.jointplot(
    data=iris,
    x="sepal_length",
    y="petal_length"
)

plt.show()
```

```python
# 13. Relplot---Ye internally relationship plots handle karta hai.

# relplot() figure-level function hai.

sns.relplot(
    data=iris,
    x="sepal_length",
    y="petal_length",
    hue="species"
)

plt.show()
```

```python
# 14. Stripplot

# Individual observations ko categorical axis par show karta hai.

sns.stripplot(
    data=iris,
    x="species",
    y="sepal_length"
)

plt.title("Sepal Length by Species")
plt.show()
```

```python
# 15. Swarmplot

# Stripplot jaisa hai, lekin points ko overlap hone se bachane ki koshish karta hai.

sns.swarmplot(
    data=iris,
    x="species",
    y="sepal_length"
)

plt.title("Sepal Length by Species")
plt.show()
```

```python
"""
16. Boxenplot

Large datasets me distribution ko boxplot se zyada detail me dekhne ke liye useful.
    """
sns.boxenplot(
    data=iris,
    x="species",
    y="sepal_length"
)

plt.title("Sepal Length Boxenplot")
plt.show()
```

```python
# 17. KDE by Species

# Species ke according petal length distribution

sns.kdeplot(
    data=iris,
    x="petal_length",
    hue="species",
    fill=True
)

plt.title("Petal Length KDE by Species")
plt.show()
```

```python
#  histogram by species 
sns.histplot(data=iris,x='petal_width',hue='species')
```

```python
#  Correlation Matrix.
corr = iris.corr(numeric_only=True)

sns.heatmap(corr,annot=True,cmap="magma")
```

```python
#  full EDA 
fig, ax = plt.subplots(2, 2, figsize=(14,10))

# 1. Species Count
sns.countplot(
    data=iris,
    x="species",
    ax=ax[0,0]
)
ax[0,0].set_title("Species Count")


# 2. Petal Length Distribution
sns.histplot(
    data=iris,
    x="petal_length",
    kde=True,
    ax=ax[0,1]
)
ax[0,1].set_title("Petal Length Distribution")


# 3. Petal Length vs Width
sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species",
    ax=ax[1,0]
)
ax[1,0].set_title("Petal Length vs Width")


# 4. Petal Length by Species
sns.boxplot(
    data=iris,
    x="species",
    y="petal_length",
    ax=ax[1,1]
)
ax[1,1].set_title("Petal Length by Species")


plt.tight_layout()
plt.show()
```
